### Motion Interpolation

Generates an "average" of 2 motions that are used as inputs

```python
# Phase 1 + Phase 2 training (toggle RUN_TRAINING to retrain)
from AE import MotionManifoldTrainer

trainer_config = dict(
    data_dir=DATA_DIR,
    output_dir=OUTPUT_DIR,
    batch_size=32,
    epochs=25,
    fine_tune_epochs=25,
    learning_rate=1e-3,
    fine_tune_lr=3e-4,
    sparsity_weight=0.01,
    window_size=160,
    val_split=0.1,
    device="cuda" if torch.cuda.is_available() else "cpu"
)

RUN_TRAINING = False  # Set to True to launch a fresh training run
if RUN_TRAINING:
    trainer = MotionManifoldTrainer(**trainer_config)
    training_stats = trainer.train()
else:
    print("Skipping training run. Set RUN_TRAINING=True to retrain the autoencoder.")
```

```python
# Interpolate motions using the trained manifold
import importlib
import AE
importlib.reload(AE)
from AE import MotionManifoldSynthesizer
from visualization import visualize_interpolation

model_path = os.path.join(OUTPUT_DIR, "models", "motion_autoencoder.pt")
if not os.path.exists(model_path):
    raise FileNotFoundError("Trained model not found. Run training before attempting interpolation.")

synthesizer = MotionManifoldSynthesizer(model_path=model_path, dataset=dataset)

total_windows = len(dataset)
interpolation_pairs = [
    (0, min(total_windows - 1, total_windows // 4)),
    (min(total_windows - 1, total_windows // 2), min(total_windows - 1, (3 * total_windows) // 4))
]

for i, (idx_a, idx_b) in enumerate(interpolation_pairs):
    if idx_a == idx_b:
        continue  # Skip degenerate pairs when the dataset is very small
    motion_a = dataset[idx_a]
    motion_b = dataset[idx_b]
    motion_a_batch = {k: v.unsqueeze(0) for k, v in motion_a.items()}
    motion_b_batch = {k: v.unsqueeze(0) for k, v in motion_b.items()}
    interpolated = synthesizer.interpolate_motions(motion_a_batch, motion_b_batch, t=0.5)
    video_path = os.path.join(VIDEO_DIR, f"interpolation_{i}.mp4")
    visualize_interpolation(
        motion_a["original_positions"].unsqueeze(0),
        motion_b["original_positions"].unsqueeze(0),
        interpolated,
        dataset.get_joint_parents(),
        video_path
    )
    print(f"Interpolation {i}: indices {idx_a} -> {idx_b}, saved to {video_path}")
    display(Video(video_path, embed=True))
```

### Fixing Corrupt Motion Data

```python
# Project corrupted motions back onto the learned manifold
from visualization import visualize_motion_comparison

corruption_settings = [
    ("zero", {"prob": 0.35}),
    ("noise", {"scale": 0.25})
]

for i, (corruption_type, params) in enumerate(corruption_settings):
    step = max(1, len(dataset) // 4)
    source_idx = min(len(dataset) - 1, (i + 1) * step)
    motion = dataset[source_idx]
    motion_batch = {k: v.unsqueeze(0) for k, v in motion.items()}
    corrupted, fixed = synthesizer.fix_corrupted_motion(motion_batch, corruption_type, params)
    video_path = os.path.join(VIDEO_DIR, f"corruption_fix_{i}.mp4")
    visualize_motion_comparison(corrupted, fixed, dataset.get_joint_parents(), video_path)
    print(f"Corruption '{corruption_type}' on sample {source_idx} saved to {video_path}")
    display(Video(video_path, embed=True))
```

## Part 4: Advanced Motion Synthesis and Editing

### Implementation

```python
import importlib
import numpy as np
import torch
from matplotlib import animation
from matplotlib import pyplot as plt
from matplotlib.animation import FFMpegWriter
import AE
import visualization
importlib.reload(AE)
importlib.reload(visualization)
from AE import MotionManifoldSynthesizer
from visualization import visualize_motion_comparison

model_path = os.path.join(OUTPUT_DIR, "models", "motion_autoencoder.pt")
extra_synth = MotionManifoldSynthesizer(model_path=model_path, dataset=dataset)
joint_parents = dataset.get_joint_parents()

def make_motion_batch(sample):
    batch = {"positions": sample["positions"].unsqueeze(0)}
    for key in ("trans_vel_xz", "rot_vel_y"):
        if key in sample:
            batch[key] = sample[key].unsqueeze(0)
    return batch

def render_style_triplet(content_motion, stylized_motion, style_motion, joint_parents, output_path, fps=30):
    """Render content, stylized, and reference style motions side-by-side."""
    if isinstance(content_motion, torch.Tensor):
        content_data = content_motion.detach().cpu().numpy()
    else:
        content_data = content_motion

    if isinstance(stylized_motion, torch.Tensor):
        stylized_data = stylized_motion.detach().cpu().numpy()
    else:
        stylized_data = stylized_motion

    if isinstance(style_motion, torch.Tensor):
        style_data = style_motion.detach().cpu().numpy()
    else:
        style_data = style_motion

    if len(content_data.shape) == 4 and content_data.shape[0] == 1:
        content_data = content_data[0]
    if len(stylized_data.shape) == 4 and stylized_data.shape[0] == 1:
        stylized_data = stylized_data[0]
    if len(style_data.shape) == 4 and style_data.shape[0] == 1:
        style_data = style_data[0]

    num_frames = min(content_data.shape[0], stylized_data.shape[0], style_data.shape[0])
    num_joints = content_data.shape[1]

    fig = plt.figure(figsize=(18, 6))
    ax_content = fig.add_subplot(131, projection='3d')
    ax_stylized = fig.add_subplot(132, projection='3d')
    ax_style = fig.add_subplot(133, projection='3d')

    ax_content.set_title('Content Motion')
    ax_stylized.set_title('Stylized Result')
    ax_style.set_title('Style Reference')

    all_data = np.concatenate([
        content_data[:num_frames],
        stylized_data[:num_frames],
        style_data[:num_frames]
    ], axis=0)
    min_vals = np.min(all_data, axis=(0, 1))
    max_vals = np.max(all_data, axis=(0, 1))
    center = (min_vals + max_vals) / 2
    max_range = np.max(max_vals - min_vals) / 2 + 0.5

    for ax in [ax_content, ax_stylized, ax_style]:
        ax.set_xlim(center[0] - max_range, center[0] + max_range)
        ax.set_ylim(center[1] - max_range, center[1] + max_range)
        ax.set_zlim(center[2] - max_range, center[2] + max_range)
        ax.set_xlabel('X')
        ax.set_ylabel('Z')
        ax.set_zlabel('Y')
        ax.view_init(elev=20, azim=45)

    def _build_lines(ax):
        lines = []
        for joint_idx in range(num_joints):
            parent_idx = joint_parents[joint_idx]
            parent_idx = parent_idx if parent_idx >= 0 else None
            if parent_idx is not None:
                line, = ax.plot([], [], [], 'k-', lw=2, alpha=0.8)
                lines.append((joint_idx, parent_idx, line))
        return lines

    lines_content = _build_lines(ax_content)
    lines_stylized = _build_lines(ax_stylized)
    lines_style = _build_lines(ax_style)

    joints_content = ax_content.scatter([], [], [], c='b', s=30)
    joints_stylized = ax_stylized.scatter([], [], [], c='g', s=30)
    joints_style = ax_style.scatter([], [], [], c='r', s=30)

    def init():
        for lines in [lines_content, lines_stylized, lines_style]:
            for _, _, line in lines:
                line.set_data([], [])
                line.set_3d_properties([])
        joints_content._offsets3d = ([], [], [])
        joints_stylized._offsets3d = ([], [], [])
        joints_style._offsets3d = ([], [], [])
        return [line for group in [lines_content, lines_stylized, lines_style] for _, _, line in group] + [
            joints_content, joints_stylized, joints_style
        ]

    def update(frame):
        ax_content.set_title(f'Content Motion (Frame {frame+1}/{num_frames})')
        ax_stylized.set_title(f'Stylized Result (Frame {frame+1}/{num_frames})')
        ax_style.set_title(f'Style Reference (Frame {frame+1}/{num_frames})')

        frame_content = content_data[frame]
        frame_stylized = stylized_data[frame]
        frame_style = style_data[frame]

        for (joint_idx, parent_idx, line) in lines_content:
            xs = [frame_content[parent_idx, 0], frame_content[joint_idx, 0]]
            ys = [frame_content[parent_idx, 2], frame_content[joint_idx, 2]]
            zs = [frame_content[parent_idx, 1], frame_content[joint_idx, 1]]
            line.set_data(xs, ys)
            line.set_3d_properties(zs)

        for (joint_idx, parent_idx, line) in lines_stylized:
            xs = [frame_stylized[parent_idx, 0], frame_stylized[joint_idx, 0]]
            ys = [frame_stylized[parent_idx, 2], frame_stylized[joint_idx, 2]]
            zs = [frame_stylized[parent_idx, 1], frame_stylized[joint_idx, 1]]
            line.set_data(xs, ys)
            line.set_3d_properties(zs)

        for (joint_idx, parent_idx, line) in lines_style:
            xs = [frame_style[parent_idx, 0], frame_style[joint_idx, 0]]
            ys = [frame_style[parent_idx, 2], frame_style[joint_idx, 2]]
            zs = [frame_style[parent_idx, 1], frame_style[joint_idx, 1]]
            line.set_data(xs, ys)
            line.set_3d_properties(zs)

        joints_content._offsets3d = (
            frame_content[:, 0], frame_content[:, 2], frame_content[:, 1]
        )
        joints_stylized._offsets3d = (
            frame_stylized[:, 0], frame_stylized[:, 2], frame_stylized[:, 1]
        )
        joints_style._offsets3d = (
            frame_style[:, 0], frame_style[:, 2], frame_style[:, 1]
        )

        return [line for group in [lines_content, lines_stylized, lines_style] for _, _, line in group] + [
            joints_content, joints_stylized, joints_style
        ]

    ani = animation.FuncAnimation(
        fig, update, frames=num_frames, init_func=init, blit=True, interval=1000/fps
    )

    writer = FFMpegWriter(fps=fps, metadata=dict(artist='Assignment 3'), bitrate=5000)
    ani.save(output_path, writer=writer)

    plt.close(fig)
    print(f"Style transfer animation saved to {output_path}")

    return output_path
```

#### Gap Filling

```python
gap_configs = [
    {"index": 35, "start": 28, "length": 48},
    {"index": len(dataset) // 5, "start": 52, "length": 40}
 ]
gap_results = []
for run_id, cfg in enumerate(gap_configs):
    sample = dataset[int(cfg["index"])]
    batch = make_motion_batch(sample)
    mask = torch.ones(sample["positions"].shape[0])
    gap_end = min(sample["positions"].shape[0], cfg["start"] + cfg["length"])
    mask[cfg["start"]:gap_end] = 0.0
    result = extra_synth.complete_motion(batch, mask.unsqueeze(0))
    video_path = os.path.join(EXTRA_DIR, f"gap_fill_{run_id}.mp4")
    visualize_motion_comparison(result["corrupted"], result["completed"], joint_parents, video_path)
    display(Video(video_path, embed=True))
    gap_results.append({"config": cfg, "video": video_path})
gap_results
```

#### Tail Extension

The main idea of this portion is similar to gap filling (where there are missing frames in the middle of the animation, and we generate new ones), but towards the end of the motion. As we can see from the examples, this is working pretty well.

```python
extension_configs = [
    {"index": len(dataset) // 3, "tail": 56},
    {"index": len(dataset) // 2, "tail": 72}
 ]
extension_results = []
for run_id, cfg in enumerate(extension_configs):
    sample = dataset[int(cfg["index"])]
    batch = make_motion_batch(sample)
    mask = torch.ones(sample["positions"].shape[0])
    sequence_length = sample["positions"].shape[0]
    tail = min(sequence_length // 2, cfg["tail"])
    if tail <= 0:
        tail = min(sequence_length, cfg["tail"])
    mask[-tail:] = 0.0
    result = extra_synth.complete_motion(batch, mask.unsqueeze(0))
    video_path = os.path.join(EXTRA_DIR, f"tail_extend_{run_id}.mp4")
    visualize_motion_comparison(result["corrupted"], result["completed"], joint_parents, video_path)
    display(Video(video_path, embed=True))
    extension_results.append({"config": cfg, "video": video_path})
extension_results
```

#### Latent Style Transfer


The main idea is that we take a normal content motion, and then change its distribution of its latent space to another "style motion" in order to give the generated motion a different style but similar motion to the original. For example, in the third generated video, the original content motion and style motion both are boxing, but the generated motion is much more laid back due to the style of the "style motion".

```python
style_pairs = [
    (15, 780, 0.65),
    (880, 1320, 0.45),
    (420, 950, 0.55)
 ]
style_results = []
for run_id, (content_idx, style_idx, alpha) in enumerate(style_pairs):
    content_sample = dataset[int(content_idx % len(dataset))]
    style_sample = dataset[int(style_idx % len(dataset))]
    content_batch = make_motion_batch(content_sample)
    style_batch = make_motion_batch(style_sample)
    transfer = extra_synth.style_transfer(content_batch, style_batch, alpha=alpha)
    video_path = os.path.join(EXTRA_DIR, f"style_transfer_{run_id}.mp4")
    render_style_triplet(transfer["content"], transfer["stylized"], transfer["style"], joint_parents, video_path)
    display(Video(video_path, embed=True))
    style_results.append({"content": int(content_idx), "style": int(style_idx), "alpha": alpha, "video": video_path})
style_results
```

## Part 5: BVH Export for Blender/Other Tools
Use the helper below to export generated motions as `.bvh` files that can be imported into Blender or other animation tools.

(Exported motions can be found in output/ae/bvh_exports)

```python
# BVH export helpers
import inspect
import importlib
import bvh_export as bvh_export_module

# Reload so notebook uses the latest on-disk exporter code.
importlib.reload(bvh_export_module)
export_motion_to_bvh = bvh_export_module.export_motion_to_bvh
EXPORT_BVH_PARAMS = set(inspect.signature(export_motion_to_bvh).parameters.keys())

BVH_DIR = os.path.join(OUTPUT_DIR, "bvh_exports")
os.makedirs(BVH_DIR, exist_ok=True)

joint_names = dataset.get_joint_names()
joint_parents = dataset.get_joint_parents()

# Canonical skeleton offsets from cached CMU data (more stable in Blender).
first_motion_key = next(iter(dataset.motion_data.keys()))
joint_offsets = np.asarray(dataset.motion_data[first_motion_key]["joint_offsets"], dtype=np.float64)


def export_bvh_motion(
    motion,
    file_name,
    fps=30,
    batch_index=0,
    generated=True,
    temporal_smoothing_alpha=0.25,
    temporal_smoothing_passes=2,
    max_angular_step_deg=30,
):
    """Export a motion tensor/dict to BVH and return the output path."""
    motion_data = motion
    if isinstance(motion, dict):
        candidate_keys = (
            "positions",
            "completed",
            "stylized",
            "content",
            "style",
            "corrupted",
            "reconstruction",
            "original_positions",
        )
        key = next((k for k in candidate_keys if k in motion), None)
        if key is None:
            raise KeyError(f"Could not find a motion tensor in dict keys: {list(motion.keys())}")
        motion_data = motion[key]

    if not file_name.endswith(".bvh"):
        file_name = f"{file_name}.bvh"

    output_path = os.path.join(BVH_DIR, file_name)

    export_kwargs = {}
    if generated:
        export_kwargs = {
            "temporal_smoothing_alpha": temporal_smoothing_alpha,
            "temporal_smoothing_passes": temporal_smoothing_passes,
            "max_angular_step_deg": max_angular_step_deg,
        }

    # Backward-compatible: only pass kwargs supported by the currently loaded exporter.
    export_kwargs = {k: v for k, v in export_kwargs.items() if k in EXPORT_BVH_PARAMS}

    base_kwargs = {
        "motion": motion_data,
        "output_path": output_path,
        "joint_names": joint_names,
        "joint_parents": joint_parents,
        "fps": fps,
        "batch_index": batch_index,
    }
    if "joint_offsets" in EXPORT_BVH_PARAMS:
        base_kwargs["joint_offsets"] = joint_offsets

    written_path = export_motion_to_bvh(
        **base_kwargs,
        **export_kwargs,
    )
    print(f"Exported: {written_path}")
    return written_path


print(f"BVH output directory: {BVH_DIR}")
print(f"Exporter signature: {inspect.signature(export_motion_to_bvh)}")
```

```python
# Auto-export motions that are currently available in memory
motions_for_export = {}

if "interpolated" in globals():
    motions_for_export["interpolation_latest"] = interpolated
if "fixed" in globals():
    motions_for_export["corruption_fixed_latest"] = fixed
if "result" in globals() and isinstance(result, dict) and "completed" in result:
    motions_for_export["completion_latest"] = result["completed"]
if "transfer" in globals() and isinstance(transfer, dict) and "stylized" in transfer:
    motions_for_export["style_transfer_latest"] = transfer["stylized"]

# Always include one baseline clip from the dataset for a quick sanity check.
baseline_idx = 0
motions_for_export[f"dataset_original_{baseline_idx}"] = dataset[baseline_idx]["original_positions"]

exported_bvh_paths = {}
for name, motion in motions_for_export.items():
    is_generated = not name.startswith("dataset_original_")
    exported_bvh_paths[name] = export_bvh_motion(
        motion,
        name,
        fps=30,
        generated=is_generated,
    )

exported_bvh_paths
```
